### Finding values

- `loc` - Label based data selecting method which means that we have to pass `the name` of the row or columnt which we want to select. 
  - df.loc[ROWS, COLUMNS]
- `iloc` - Indexed-based selecting method which means that we have to pass an `integer index` in the method to select a specific row/column
  - df.iloc[ROWS, COLUMNS] # But only using integer index

## Adding columns

### Basic add

```python
bios_copy = bios.copy()
```

```python
bios_copy["first_name"] = bios["name"].str.split(" ").str[0]
bios_copy.head()
```

### Datetime

```python
bios["born_datetime"] = pd.to_datetime(bios["born_date"], format="%Y-%m-%d")
```

```python
bios_copy["born_year"] = bios["born_datetime"].dt.year
bios_copy.head()
```

### Apply

```python
bios_copy["height_category"] = bios["height_cm"].apply(
    lambda x: "Short" if x < 165 else ("Average" if x < 185 else "Tall")
)
bios_copy.head()
```

```python
def categorize_athlete(row):
    if row["height_cm"] <= 175 or row["weight_kg"] <= 75:
        return "Lightweight"
    if row["height_cm"] <= 185 or row["weight_kg"] <= 85:
        return "Middleweight"

    return "Heavyweight"
```

```python
bios_copy["heavy_category"] = bios.apply(lambda x: categorize_athlete(x), axis=1)
bios_copy.head()
```

## Shift | Rank | Rolling | Cumsum

### Shift

Shift index by desired number of periods with an optional time freq.

```python
coffee.head()
```

```python
coffee["Yesterday_Revenue"] = coffee['Revenue'].shift(2)
coffee.head()
```

```python
coffee['PCT_Change'] = coffee['Revenue'] / coffee['Yesterday_Revenue'] * 100
coffee.head()
```

### Rank

Compute numerical data ranks (1 through n) along axis.

```python
bios['height_rank'] = bios['height_cm'].rank(ascending=False)
bios.sort_values('height_rank')
```

### Cumsum
