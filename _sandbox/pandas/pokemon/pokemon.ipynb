# Changes in data

Making changes in data can be done in multiple ways.

```python
df.head(1)
```

```python
# Naive approach
df["Total"] = (
    df["HP"]
    + df["Attack"]
    + df["Defense"]
    + df["Sp. Atk"]
    + df["Sp. Def"]
    + df["Speed"]
)

df.head()
```

```python
df["Total"] = df.loc[:, ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]].sum(
    axis=1
)

df.head()
```

```python
df["Total"] = df.iloc[:, 4:10].sum(axis=1)
df.head()
```

```python
features = ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]
df["Total"] = df[features].apply(lambda row: row.sum(), axis=1)
df.head()
```

```python
features = ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]
df["Total"] = df[features].sum(axis=1)

df.head()
```

# Filtering data

## Filtering by specific column

```python
df.loc[
    (df["Type 1"] == "Grass") & (df["Type 2"] == "Poison") & (df["HP"] >= 70)
].sort_values("Name")
```

```python
df.loc[df["Legendary"]].sort_values(["HP", "Total"])
```

## Containing phrase

Grab every pokemon with Mega in name.

```python
df.loc[df["Name"].str.contains("Mega")].head()
```

```python
df["Name"] = df["Name"].str.replace(r"(\w+)(Mega)\s([a-zA-z]+)", r"\1 \2", regex=True)
```

```python
df.loc[df["Name"].str.contains("Mega")].tail()
```

# Aggregate Statistics (GroupBy)

## Grass Pokemon - average HP

```python
df.head()
```

```python
df.groupby(by="Type 1")[
    ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]
].mean().sort_values(["Sp. Def"])
```

```python
(df.groupby(by="Type 1")["Name"].count() / len(df) * 100).to_frame(
    name="Percentage"
).sort_values(by="Percentage", ascending=True)
```
