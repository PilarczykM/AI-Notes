```python
from audiocraft.models import MusicGen
from audiocraft.data.audio import audio_write

from IPython.display import Audio
```

```python
class CFG:
    model = 'nateraw/musicgen-songstarter-v0.2'
    duration = 15
```

# Model

```python
model = MusicGen.get_pretrained(CFG.model)
```

```python
descriptions = ['melodic heavy metal', 'electric blues', 'symphonic movie soundtrack']

wav = model.generate(descriptions, progress=True)
```

```python
for idx, one_wav in enumerate(wav):
    # Will save under {idx}.wav, with loudness normalization at -14 db LUFS.
    audio_write(f'{idx}', one_wav.cpu(), model.sample_rate, strategy="loudness", loudness_compressor=True)
```

## Promptujemy melodią

```python
from huggingface_hub import hf_hub_download
import torchaudio
import librosa
```

```python
melody_path = hf_hub_download(CFG.model, 'assets/bach.mp3')
Audio(melody_path)
```

```python
melody, sr = torchaudio.load(melody_path)

duration = librosa.get_duration(y=melody, sr=sr)
print(duration)
model.set_generation_params(duration = duration)

model.generate_with_chroma(
    descriptions=descriptions,
    melody_wavs=melody[None].expand(len(descriptions), -1, -1),
    melody_sample_rate=sr)

for idx, one_wav in enumerate(wav):
    # Will save under {idx}.wav, with loudness normalization at -14 db LUFS.
    audio_write(f'{idx}_bach', one_wav.cpu(), model.sample_rate, strategy="loudness", loudness_compressor=True)
```
