- `duckduckgo-search` - biblioteka umożliwiająca przeprowadzanie wyszukiwań w internecie za pomocą wyszukiwarki DuckDuckGo

````python
from huggingface_hub import login
from google.colab import userdata
from smolagents import load_tool, CodeAgent, HfApiModel, DuckDuckGoSearchTool
````

````python
class CFG:
    muhtool = "m-ric/text-to-image"
    model = "Qwen/Qwen2.5-72B-Instruct"
````

````python
login(userdata.get("HF_TOKEN"))
````

# Model

````python
image_generation_tool = load_tool(CFG.muhtool, trust_remote_code=True)

search_tool = DuckDuckGoSearchTool()
````

````python
model = HfApiModel(CFG.model)

agent = CodeAgent(tools=[image_generation_tool, search_tool], model=model)
````

````python
result = agent.run(
    "Generate me a photo of the car that James Bond drove in the latest movie.",
)
result
````

Ta linijka kodu uruchamia agenta AI do wykonania zadania oraz wyświetla jego rezultat:

```python
result = agent.run(
    "Generate me a photo of the car that James Bond drove in the latest movie.",
)
result
```

Przeanalizujmy ją krok po kroku:

Funkcja `agent.run()` nakazuje wcześniej skonfigurowanemu agentowi wykonanie zadania określonego przez tekst przekazany jako argument. W tym przypadku zadanie to: "Wygeneruj mi zdjęcie samochodu, którym James Bond jeździł w najnowszym filmie."

Gdy agent otrzymuje to polecenie, uruchamiana jest cała złożona sekwencja operacji:

1. Model językowy Qwen 2.5 analizuje zapytanie i rozumie, że chodzi o wygenerowanie obrazu samochodu Jamesa Bonda.

2. Ponieważ to zadanie wymaga aktualnej wiedzy (agent musi wiedzieć, jaki samochód pojawił się w najnowszym filmie o Bondzie), agent prawdopodobnie użyje narzędzia `search_tool` (DuckDuckGo), aby wyszukać informacje o najnowszym filmie o Jamesie Bondzie i samochodzie, który w nim wystąpił.

3. Po znalezieniu informacji o konkretnym modelu samochodu, agent użyje narzędzia `image_generation_tool` do stworzenia obrazu tego samochodu. Może to być np. Aston Martin (tradycyjnie kojarzony z Bondem) lub inny ekskluzywny model, zależnie od tego, co faktycznie pojawiło się w najnowszym filmie.

4. Rezultat działania (wygenerowany obraz) zostaje zapisany w zmiennej `result`.

5. Ostatnia linijka `result` powoduje wyświetlenie zawartości tej zmiennej w środowisku interaktywnym (np. Jupyter Notebook lub Google Colab).

To, co zobaczymy jako wynik, to najprawdopodobniej wygenerowany obraz samochodu lub strukturę danych zawierającą zarówno obraz, jak i dodatkowe informacje, takie jak model samochodu i film, z którego pochodzi.

Jest to doskonały przykład integracji różnych zdolności AI w jednym systemie:
- Zdolność rozumienia języka naturalnego (interpretacja zapytania)
- Zdolność wyszukiwania aktualnych informacji (znalezienie właściwego modelu samochodu)
- Zdolność generowania obrazów (stworzenie wizualizacji samochodu)

Ten kod demonstruje, jak zaawansowane systemy AI mogą współdziałać z internetem i narzędziami zewnętrznymi, aby dostarczać wyniki, które wykraczają poza możliwości pojedynczego modelu.
