```python
from any_agent import AgentConfig, AnyAgent, TracingConfig
from any_agent.tools import search_web
from any_agent.evaluation.evaluation_case import EvaluationCase
from any_agent.evaluation import evaluate, EvaluationCase

import os
from google.colab import userdata

import nest_asyncio

nest_asyncio.apply()
```

Ten kod importuje różne elementy z biblioteki `any-agent` oraz inne potrzebne moduły. Przygotowuje środowisko do korzystania z tej biblioteki, szczególnie w kontekście tworzenia i ewaluacji agentów AI.

Rozbijmy to na poszczególne linie:

*   `from any_agent import AgentConfig, AnyAgent, TracingConfig`: Importuje trzy klasy z modułu `any_agent`.
    *   `AgentConfig`: Klasa służąca do konfigurowania agenta AI. Określa parametry takie jak model językowy, temperatura itp.
    *   `AnyAgent`: Główna klasa reprezentująca agenta AI.  Zawiera logikę działania agenta i interakcji z narzędziami.
    *   `TracingConfig`: Klasa służąca do konfigurowania śledzenia (tracing) działań agenta, co pozwala na analizę jego zachowania.

*   `from any_agent.tools import search_web`: Importuje funkcję `search_web` z modułu `any_agent.tools`. Ta funkcja umożliwia agentowi wyszukiwanie informacji w internecie.

*   `from any_agent.evaluation.evaluation_case import EvaluationCase`: Importuje klasę `EvaluationCase` z modułu `any_agent.evaluation.evaluation_case`.  Ta klasa reprezentuje pojedynczy przypadek testowy używany do oceny agenta.

*   `from any_agent.evaluation import evaluate, EvaluationCase`: Ponownie importuje `EvaluationCase` (może to być celowe dla jasności lub redundancja) oraz funkcję `evaluate` z modułu `any_agent.evaluation`. Funkcja `evaluate` służy do przeprowadzania ewaluacji agenta na podstawie zestawu przypadków testowych.

```python
class CFG:
    model = "gpt-4o-mini"

os.environ["OPENAI_API_KEY"] = userdata.get("openaivision")
```

# Test

```python
agent = AnyAgent.create(
    "langchain",
    AgentConfig(model_id=CFG.model, tools=[search_web]),
    tracing=TracingConfig(console=True, cost_info=False),
)
```

```python
agent_trace = agent.run(
    "How many seconds would it take for a leopard at full speed to run through Pont des Arts?"
)
```

```python
evaluation_case = EvaluationCase(
    ground_truth=[{"name": "Seconds", "value": "9", "points": 1.0}],
    checkpoints=[
        {"criteria": "Did the agent run a calculation", "points": 1},
        {"criteria": "Did the agent use fewer than 5 steps", "points": 4},
    ],
    llm_judge=CFG.model,
)
```

Ten kod tworzy instancję klasy `EvaluationCase`, definiując przypadek testowy do oceny działania agenta AI. Określa on oczekiwaną odpowiedź (ground truth), kryteria oceny oraz model językowy używany jako sędzia do weryfikacji odpowiedzi agenta.

Rozbijmy to na części:

*   `evaluation_case = EvaluationCase(...)`: Tworzy nową instancję klasy `EvaluationCase`, która reprezentuje pojedynczy przypadek testowy.
*   `ground_truth=[{"name": "Seconds", "value": "9", "points": 1.0}]`: Definiuje oczekiwaną, prawidłową odpowiedź na zadane pytanie. Jest to lista słowników, gdzie każdy słownik reprezentuje jeden element odpowiedzi.
    *   `"name": "Seconds"`: Nazwa pola w odpowiedzi (w tym przypadku czas podany w sekundach).
    *   `"value": "9"`: Oczekiwana wartość dla tego pola (9 sekund).
    *   `"points": 1.0`: Maksymalna liczba punktów, jaką można uzyskać za prawidłową odpowiedź na to pole.
*   `checkpoints=[...]`: Definiuje kryteria oceny odpowiedzi agenta, niezależnie od dokładnej wartości. Są to punkty kontrolne, które sprawdzają, czy agent wykonał określone kroki lub spełnił pewne warunki.
    *   `{"criteria": "Did the agent run a calculation", "points": 1}`: Sprawdza, czy agent przeprowadził obliczenia (np. przeliczył prędkość leoparda i długość mostu). Jeśli tak, przyznawane są 1 punkt.
    *   `{"criteria": "Did the agent use fewer than 5 steps", "points": 4}`: Sprawdza, czy agent rozwiązał problem w mniej niż 5 krokach (np. wywołań narzędzi lub iteracji). Jeśli tak, przyznawane są 4 punkty.
*   `llm_judge = CFG.model`: Określa model językowy, który będzie używany jako "sędzia" do oceny odpowiedzi agenta. W tym przypadku jest to ten sam model co w agencie (`'gpt-4o-mini'`), zdefiniowany w klasie `CFG`. Model sędzia porównuje odpowiedź agenta z oczekiwaną odpowiedzią (ground truth) i sprawdza, czy spełnione zostały kryteria oceny.

Podsumowując, ten kod definiuje przypadek testowy, który ma na celu ocenę zdolności agenta do rozwiązywania problemu obliczeniowego w efektywny sposób. Oczekiwana odpowiedź to 9 sekund, a dodatkowe punkty są przyznawane za przeprowadzenie obliczeń i rozwiązanie problemu w niewielkiej liczbie kroków. Model językowy GPT-4o-mini będzie pełnił rolę sędziego oceniającego odpowiedź agenta.

```python
eval_result = evaluate(
    evaluation_case=evaluation_case,
    trace=agent_trace,
    agent_framework="OPENAI",
)
print(f"Final score: {eval_result.score}")
```

Ten kod przeprowadza ewaluację działania agenta AI na podstawie wcześniej zdefiniowanego przypadku testowego i śladu jego działania, a następnie wyświetla wynik ewaluacji.

Rozbijmy to na części:

*   `eval_result = evaluate(...)`: Wywołuje funkcję `evaluate`, która przeprowadza ocenę odpowiedzi agenta w oparciu o podane argumenty. Funkcja ta zwraca obiekt zawierający wyniki ewaluacji.
    *   `evaluation_case=evaluation_case`: Przekazuje instancję klasy `EvaluationCase`, czyli zdefiniowany wcześniej przypadek testowy, który zawiera oczekiwaną odpowiedź i kryteria oceny.
    *   `trace=agent_trace`: Przekazuje ślad działania agenta (`agent_trace`), który został zapisany podczas uruchomienia agenta z zapytaniem. Ślad ten zawiera informacje o tym, jakie kroki podjął agent, jakie narzędzia wykorzystał i jaką odpowiedź wygenerował.
    *   `agent_framework="OPENAI"`: Określa framework, którego używał agent. W tym przypadku jest to "OPENAI", co sugeruje, że agent korzysta z API OpenAI.
*   `print(f"Final score: {eval_result.score}")`: Wyświetla końcowy wynik ewaluacji (score) na konsoli. `eval_result.score` to atrybut obiektu `eval_result`, który zawiera sumę punktów uzyskanych przez agenta za spełnienie kryteriów oceny i podanie prawidłowej odpowiedzi.

W efekcie, ten kod uruchamia proces ewaluacji, porównuje odpowiedź agenta z oczekiwaną odpowiedzią, sprawdza spełnienie kryteriów oceny i wyświetla końcowy wynik ewaluacji, który wskazuje na jakość działania agenta w danym przypadku testowym.
